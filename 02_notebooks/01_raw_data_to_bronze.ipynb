### Ingest Raw Data from Volume to Bronze layer
- This is copy activity of the raw data as-is into the bronze schema as a delta table

```python
%run ./00_config
```

#### Create rat_sighting_bronze table

```python
%sql
-- Create the rat_sightings_bronze table
CREATE OR REPLACE TABLE rat_sightings_bronze
AS
SELECT *
FROM read_files(
  '/Volumes/nyc_rats/source_data/raw_data/rat_sightings.csv',
  format                   => 'csv',
  header                   => 'true',
  inferSchema              => 'true',
  delimiter                => ','
);

-- Display the table
SELECT *
FROM rat_sightings_bronze
LIMIT 10;
```

#### Restaurant Inspections

```python
%sql
-- Create the rat_sightings_bronze table
CREATE OR REPLACE TABLE restaurant_inspections_bronze
AS
SELECT 
  *,
  _metadata.file_path AS file_path,
  _metadata.file_modification_time AS file_modification_time,
  _metadata.input_file_name AS file_name
FROM read_files(
  '/Volumes/nyc_rats/source_data/raw_data/restaurant_inspections.csv',
  format                   => 'csv',
  header                   => 'true',
  inferSchema              => 'true',
  delimiter                => ','
);

-- Display the table
SELECT *
FROM restaurant_inspections_bronze
LIMIT 10;
```
