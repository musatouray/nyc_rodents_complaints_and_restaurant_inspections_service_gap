```python
# Imports
from pyspark.sql import functions as F

# --- Environment ---------------------------------------------------------
dbutils.widgets.text("env", "dev")
ENV = dbutils.widgets.get("env")

# Environment -> catalog.
CATALOG_BY_ENV = {"dev": "nyc_rats", "prod": "nyc_rats_prod"}
MY_CATALOG = CATALOG_BY_ENV.get(ENV, "nyc_rats")

# Medallion schema names
BRONZE, SILVER, GOLD = "01_bronze", "02_silver", "03_gold"

# Set the active catalog; default schema is bronze
# it with use_schema(SILVER) / use_schema(GOLD) right after %run).
spark.sql(f"USE CATALOG {MY_CATALOG}")
spark.sql(f"USE SCHEMA `{BRONZE}`")

def use_schema(schema: str):
    """Set the default schema for this notebook's layer."""
    spark.sql(f"USE SCHEMA `{schema}`")
    print(f"catalog = {MY_CATALOG} (env={ENV}) | default schema = {schema}")

print(f"catalog = {MY_CATALOG} (env={ENV}) | default schema = {BRONZE}")
```
